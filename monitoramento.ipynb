```python

import time

def extrair():
    time.sleep(1)
    return [100, 200, 300]

def transformar(dados):
    time.sleep(1)
    return [d * 1.1 for d in dados]

def carregar(dados):
    time.sleep(1)
    print("Dados carregados:", dados)
```

```python
import logging
import time

logging.basicConfig(
    filename="pipeline.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def monitorar_pipeline():
    inicio = time.time()
    logging.info("Pipeline iniciado")

    try:
        dados = extrair()
        logging.info("Extração concluída")

        dados = transformar(dados)
        logging.info("Transformação concluída")

        carregar(dados)
        logging.info("Carga concluída")

    except Exception as e:
        logging.error(f"Erro no pipeline: {e}")

    fim = time.time()
    duracao = fim - inicio

    logging.info(f"Pipeline finalizado em {duracao:.2f} segundos")
```

```python
monitorar_pipeline()
```

```python
import logging
import time
import uuid
from datetime import datetime


# ==========================================
# CONFIGURAÇÃO DO LOGGER
# ==========================================

logger = logging.getLogger("pipeline_monitoramento")
logger.setLevel(logging.INFO)

formatter = logging.Formatter(
    "%(asctime)s | %(levelname)s | %(message)s"
)

# Handler para arquivo
file_handler = logging.FileHandler("pipeline.log", encoding="utf-8")
file_handler.setLevel(logging.INFO)
file_handler.setFormatter(formatter)

# Handler para console
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(formatter)

# Evita duplicação de handlers
if not logger.handlers:
    logger.addHandler(file_handler)
    logger.addHandler(console_handler)


# ==========================================
# FUNÇÕES DE EXEMPLO
# ==========================================

def extrair():
    time.sleep(1)

    dados = [
        {"id": 1, "nome": "Ana", "valor": 100},
        {"id": 2, "nome": "Bruno", "valor": 200},
        {"id": 3, "nome": "Carla", "valor": 300},
    ]

    return dados


def transformar(dados):
    time.sleep(1)

    dados_transformados = []
    for registro in dados:
        novo = registro.copy()
        novo["valor_com_imposto"] = round(registro["valor"] * 1.1, 2)
        dados_transformados.append(novo)

    return dados_transformados


def carregar(dados):
    time.sleep(1)

    # Simula carga
    quantidade = len(dados)
    return quantidade


# ==========================================
# FUNÇÃO AUXILIAR PARA LOG DE ETAPA
# ==========================================

def log_etapa(nome_etapa, inicio_etapa, quantidade=None, status="SUCESSO"):
    fim_etapa = time.time()
    duracao = fim_etapa - inicio_etapa

    mensagem = (
        f"etapa={nome_etapa} | status={status} | duracao_segundos={duracao:.2f}"
    )

    if quantidade is not None:
        mensagem += f" | quantidade_registros={quantidade}"

    logger.info(mensagem)


# ==========================================
# MONITORAMENTO COMPLETO DO PIPELINE
# ==========================================

def monitorar_pipeline():
    pipeline_id = str(uuid.uuid4())
    inicio_pipeline = time.time()
    data_execucao = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    logger.info("=" * 100)
    logger.info(
        f"pipeline_id={pipeline_id} | evento=INICIO_PIPELINE | data_execucao={data_execucao}"
    )

    try:
        # ------------------------------
        # EXTRAÇÃO
        # ------------------------------
        inicio_extracao = time.time()
        logger.info(f"pipeline_id={pipeline_id} | etapa=EXTRACAO | evento=INICIO")

        dados_extraidos = extrair()

        log_etapa(
            nome_etapa="EXTRACAO",
            inicio_etapa=inicio_extracao,
            quantidade=len(dados_extraidos),
            status="SUCESSO"
        )

        # ------------------------------
        # TRANSFORMAÇÃO
        # ------------------------------
        inicio_transformacao = time.time()
        logger.info(f"pipeline_id={pipeline_id} | etapa=TRANSFORMACAO | evento=INICIO")

        dados_transformados = transformar(dados_extraidos)

        log_etapa(
            nome_etapa="TRANSFORMACAO",
            inicio_etapa=inicio_transformacao,
            quantidade=len(dados_transformados),
            status="SUCESSO"
        )

        # ------------------------------
        # CARGA
        # ------------------------------
        inicio_carga = time.time()
        logger.info(f"pipeline_id={pipeline_id} | etapa=CARGA | evento=INICIO")

        quantidade_carregada = carregar(dados_transformados)

        log_etapa(
            nome_etapa="CARGA",
            inicio_etapa=inicio_carga,
            quantidade=quantidade_carregada,
            status="SUCESSO"
        )

        status_final = "SUCESSO"

    except Exception as e:
        status_final = "ERRO"
        logger.exception(
            f"pipeline_id={pipeline_id} | evento=ERRO_PIPELINE | mensagem={str(e)}"
        )

    fim_pipeline = time.time()
    duracao_total = fim_pipeline - inicio_pipeline

    logger.info(
        f"pipeline_id={pipeline_id} | evento=FIM_PIPELINE | status={status_final} | duracao_total_segundos={duracao_total:.2f}"
    )
    logger.info("=" * 100)


# ==========================================
# EXECUÇÃO
# ==========================================

if __name__ == "__main__":
    monitorar_pipeline()
```
