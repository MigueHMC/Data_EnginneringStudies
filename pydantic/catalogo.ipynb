```python
from pydantic import BaseModel, Field, EmailStr

class Cliente(BaseModel):
    cliente_id: int = Field(description="Identificador único do cliente")
    nome: str = Field(description="Nome completo do cliente")
    email: EmailStr = Field(description="E-mail do cliente")
```

```python
cliente = Cliente(
    cliente_id=1,
    nome="Ana Souza",
    email="ana@email.com"
)

print(cliente)
```

```python
cliente = Cliente(
    cliente_id=2,
    nome="Miguel",
    email="miguelhmc@email.com"
)

print(cliente)
```

```python
import pandas as pd

df = pd.DataFrame({
    "cliente_id": [1, 2, 3],
    "nome": ["Ana Souza", "Bruno Lima", "Carla Dias"],
    "email": ["ana@email.com", "bruno@email.com", "carla@email.com"]
})

print(df)
```

```python
import great_expectations as gx

context = gx.get_context()
```

#Validador

```python
datasource = context.data_sources.add_pandas("pandas_source")

data_asset = datasource.add_dataframe_asset(
    name="clientes_asset"
)

batch_definition = data_asset.add_batch_definition_whole_dataframe(
    "clientes_batch"
)

batch = batch_definition.get_batch(
    batch_parameters={"dataframe": df}
)

validator = context.get_validator(batch=batch)
```

```python
validator.expect_column_values_to_not_be_null("email")
validator.expect_column_values_to_not_be_null("nome")
validator.expect_column_values_to_be_unique("cliente_id")
```

#Resultado

```python
resultado = validator.validate()

print(resultado)
```
