```python
from pydantic import BaseModel, EmailStr, Field
from typing import Optional

class Cliente(BaseModel):
    cliente_id: int
    nome: str
    idade: int = Field(gt=17)
    email: EmailStr
    saldo: Optional[float] = 0
```

```python
cliente_valido = Cliente(
    cliente_id=1,
    nome="Ana",
    idade=30,
    email="ana@email.com",
    saldo=1000
)

print(cliente_valido)
```
